## Visualizing Core Findings
We have pre-generated high-quality visuals for our presentation. We can inspect the fundamental business drivers: Contract type and Tenure.

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

def show_image(image_filename):
    path = os.path.join('../visuals', image_filename)
    if os.path.exists(path):
        plt.figure(figsize=(12,8))
        img = mpimg.imread(path)
        plt.imshow(img)
        plt.axis('off')
        plt.show()
    else:
        print(f"File not found: {path}")

show_image('01_churn_distribution.png')
show_image('02_churn_by_contract.png')
show_image('03_churn_by_tenure.png')
```
